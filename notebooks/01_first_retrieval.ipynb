```python
from pathlib import Path

PROJECT_ROOT = Path("/content/college-knowledge-assistant-sync")
document_path = (
    PROJECT_ROOT / "data" / "sample_documents" / "academic_regulations.txt"
)

document = document_path.read_text(encoding="utf-8")

print(document)
```

```python
def split_into_chunks(text: str) -> list[str]:
    """Split text into non-empty paragraphs."""
    return [
        paragraph.strip()
        for paragraph in text.split("\n\n")
        if paragraph.strip()
    ]


chunks = split_into_chunks(document)

print(f"Number of chunks: {len(chunks)}")

for index, chunk in enumerate(chunks, start=1):
    print(f"\n--- Chunk {index} ---")
    print(chunk)
```

```python
def retrieve_by_keyword(
    question: str,
    chunks: list[str],
    top_k: int = 2,
) -> list[str]:
    """Rank chunks by the number of question keywords they contain."""
    keywords = set(question.lower().split())

    scored_chunks = []

    for chunk in chunks:
        words = set(chunk.lower().replace(".", "").replace(",", "").split())
        score = len(keywords & words)
        scored_chunks.append((score, chunk))

    scored_chunks.sort(key=lambda item: item[0], reverse=True)

    return [
        chunk
        for score, chunk in scored_chunks[:top_k]
        if score > 0
    ]


question = "What attendance percentage is required for examinations?"

results = retrieve_by_keyword(question, chunks)

print("Question:", question)

for index, result in enumerate(results, start=1):
    print(f"\nResult {index}:\n{result}")
```
